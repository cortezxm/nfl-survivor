# Weekly report: moneyline coverage
For the selected week, which games already have moneylines in the snapshot, and which are still missing?

```python
from pathlib import Path
SNAPSHOT_DIR = Path("../data/snapshots")
SNAPSHOT = sorted(SNAPSHOT_DIR.glob("schedule_*.csv"))[-1]
SEASON = 2026
WEEK = 5

print(SNAPSHOT)
```

```python
import pandas as pd

games = pd.read_csv(SNAPSHOT)
games.shape
```

```python
week_games = games[(games["season"] == SEASON) & (games["week"] == WEEK)]
week_games.shape
```

```python
week_games["has_line"] = week_games["home_moneyline"].notna() & week_games["away_moneyline"].notna() 
week_games[["gameday", "away_team", "home_team", "away_moneyline", "home_moneyline", "has_line"]]
```

```python
week_games["has_line"].value_counts()
```
